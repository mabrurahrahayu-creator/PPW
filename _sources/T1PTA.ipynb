Teknik Industri

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

def Industri():
    data = {
        "penulis": [],
        "judul": [],
        "pembimbing_pertama": [],
        "pembimbing_kedua": [],
        "abstrak_indo": [],
        "abstrak_inggris": [],
        "prodi" : []
    }

    for i in range(1, 144):
        print(f"📄 Memproses halaman ke-{i}")
        url = "https://pta.trunojoyo.ac.id/c_search/byprod/9/{}".format(i)
        r = requests.get(url)
        request = r.content
        soup = BeautifulSoup(request, "html.parser")
        jurnals = soup.select('li[data-cat="#luxury"]')

        for jurnal in jurnals:
            response = requests.get(jurnal.select_one('a.gray.button')['href'])
            soup1 = BeautifulSoup(response.content, "html.parser")

            isi = soup1.select_one('div#content_journal')

            judul = isi.select_one('a.title').text.strip()
            penulis = isi.find('span', string=lambda t: t and "Penulis" in t).text.split(' : ')[1].strip()
            pembimbing_pertama = isi.find('span', string=lambda t: t and "Dosen Pembimbing I" in t).text.split(' : ')[1].strip()
            pembimbing_kedua = isi.find('span', string=lambda t: t and "Dosen Pembimbing II" in t).text.split(' :')[1].strip()

            abstrak_paragraf = isi.find_all('p', align="justify")
            abstrak_indo = abstrak_paragraf[0].get_text(strip=True) if len(abstrak_paragraf) >= 1 else ''
            abstrak_inggris = abstrak_paragraf[1].get_text(strip=True) if len(abstrak_paragraf) >= 2 else ''

            data["penulis"].append(penulis)
            data["judul"].append(judul)
            data["pembimbing_pertama"].append(pembimbing_pertama)
            data["pembimbing_kedua"].append(pembimbing_kedua)
            data["abstrak_indo"].append(abstrak_indo)
            data["abstrak_inggris"].append(abstrak_inggris)
            data["prodi"].append("Teknik Industri")

        time.sleep(1)  # delay 1 detik tiap halaman

    df = pd.DataFrame(data)
    df.to_csv("Industri.csv", index=False)

    return df
```
