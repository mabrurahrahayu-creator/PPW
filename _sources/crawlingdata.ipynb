```python
import requests
import pandas as pd

api_key = "b00d4a5f35130791d9701c6257733011"

keywords = ["web mining", "web usage mining"]

all_results = []

for keyword in keywords:
    url = "https://api.springernature.com/openaccess/json"  # pakai endpoint openaccess
    params = {
        "q": keyword,
        "api_key": api_key,
        "p": 10
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        data = response.json()
        print(f"\n🔎 Keyword: {keyword}")
        print(f"Total hasil: {data['result'][0]['total']}\n")

        for record in data['records']:
            doi = record.get('doi', 'N/A')
            title = record.get('title', 'No title')
            abstract = record.get('abstract', 'No abstract')

            all_results.append({
                "keyword": keyword,
                "doi": doi,
                "title": title,
                "abstract": abstract
            })

            print(f"DOI: {doi}")
            print(f"Title: {title}\n")
    else:
        print("❌ Error:", response.status_code, response.text)

df = pd.DataFrame(all_results)
df.to_csv("hasil_crawling.csv", index=False)
print("\n✅ Data berhasil disimpan ke springer_results.csv")
```
